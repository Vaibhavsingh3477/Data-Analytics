```python
# read a text file and count the number of lines, words, and characters.
def count_file_contents(filename):
    with open(filename, 'r') as f:
        content = f.read()
        lines = content.splitlines()
        words = content.split()
        characters = len(content)
        
        print(f"Lines: {len(lines)}")
        print(f"Words: {len(words)}")
        print(f"Characters: {characters}")
count_file_contents('example.txt')

```

```python
# writing and the reading a file

with open('example.txt', 'w+') as file:
    file.write("This is a sample text file.\n")
    file.write("It contains multiple lines of text.\n")
    file.write("This is the last line.")

    # move the file cursor to the begining
    file.seek(0)
    content = file.read()
    print(content)
```
